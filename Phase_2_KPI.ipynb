```python
## Converted timestamp column

web_combined['date_time'] = pd.to_datetime(web_combined['date_time'])

web_combined['date_time']
```

```python
## Sorting Data, ensuring steps are in correct order

web_combined = web_combined.sort_values(
    by=['client_id', 'date_time']
)
web_combined
```

```python
## Calculate next time stamp to get next step timestamp

web_combined['next_time'] = web_combined.groupby('client_id')['date_time'].shift(-1)

web_combined
```

```python
## Calculating time spent(every row will contain time spent on that step)

web_combined['time_spent_sec'] = (
    web_combined['next_time'] - web_combined['date_time']
).dt.total_seconds()

web_combined['time_spent_sec']
```

```python
## Removing Odd values
## negative duration
## inactive sessions above 30 mins

web_combined = web_combined[
    (web_combined['time_spent_sec'] >= 0) &
    (web_combined['time_spent_sec'] <= 1800)
]

web_combined
```

```python
## KPI Calculation

## Average time per step: mean

step_kpi = web_combined.groupby('process_step')['time_spent_sec'].mean()

step_kpi
```

```python
## Average time per step: Median

step_median = web_combined.groupby('process_step')['time_spent_sec'].median()
step_median
```

```python
## Comparing New VS Old design
## IMPORTANT for redesign analysis
## can compare control vs step
## can compare step by step


variat_kpi = web_combined.groupby(
    ['Variation', 'process_step']
)['time_spent_sec'].mean().reset_index()

variat_kpi
```

```python
## Hypothesis Test 1
## Average Time Spent per Step (Control vs Test)
```

```python
web_combined[['Variation_x','Variation_y','Variation']].head()
```

```python
web_combined['Variation'].value_counts(dropna=False)
```

```python
## Created two samples control_times and test_times
```

```python
control_times = web_combined[
    web_combined['Variation'] == 'Control'
]['time_spent_sec']

test_times = web_combined[
    web_combined['Variation'] == 'Test'
]['time_spent_sec']

print("Control:", len(control_times))
print("Test:", len(test_times))
```

```python
## Average time in each group
```

```python
print("Control Mean:", control_times.mean())
print("Test Mean:", test_times.mean())
```

```python
## H0: The average time spent per step is the same in Control and Test.
## H0: μControl= μTest
```

```python
## The average time spent per step is different in Control and Test.
## H1:μControl != μTest
```

```python
## Statistical test(t-test)

from scipy.stats import ttest_ind

t_stat, p_value = ttest_ind(
    control_times,
    test_times,
    equal_var=False,
    nan_policy='omit'
)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
## Hypothesis 1: Average Time Spent per Step
```

```python
## H₀: There is no difference in the average time spent per step between users in the Control and Test groups.
```

```python
## H₁: There is a difference in the average time spent per step between users in the Control and Test groups.
```

```python
## An independent Welch's t-test was conducted to compare the average time spent per step between the Control and Test groups. The test produced a t-statistic of -0.898 and a p-value of 0.369.
```

```python
## Since the p-value (0.369) is greater than the significance level of 0.05, we fail to reject the null hypothesis.
```

```python
## The analysis does not provide sufficient statistical evidence that the redesigned process changed the average time users spent on steps compared to the original design. Any observed differences are likely due to random variation rather than the redesign itself.
```

```python
## Hypothesis 2 (confirm step duration)
```

```python
## Step-specific Hypothesis(The chart showed a noticeable difference at the confirm step.)
```

```python
## Created a Confirm Step samples
```

```python
confirm_control = web_combined[
    (web_combined['Variation'] == 'Control') &
    (web_combined['process_step'] == 'confirm')
]['time_spent_sec']

confirm_test = web_combined[
    (web_combined['Variation'] == 'Test') &
    (web_combined['process_step'] == 'confirm')
]['time_spent_sec']

print("Control Confirm:", len(confirm_control))
print("Test Confirm:", len(confirm_test))

print("Control Mean:", confirm_control.mean())
print("Test Mean:", confirm_test.mean())
```

```python
## T-test
```

```python
from scipy.stats import ttest_ind

t_stat, p_value = ttest_ind(
    confirm_control,
    confirm_test,
    equal_var=False,
    nan_policy='omit'
)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
## Null Hypothesis (H₀): There is no significant difference in the average time spent on the confirm step between the Control and Test groups.
## H0 : μconfirm,Control = μconfirm,Test
```

```python
## Alternative Hypothesis (H₁):
## There is a significant difference in the average time spent on the confirm step between the Control and Test groups.
## H1 : μconfirm,Control != μconfirm,Test
```

```python
## Statistical Test:
```

```python
## An independent Welch's t-test was performed to compare the average time spent on the confirm step between users in the Control and Test groups.
## The analysis produced:
## Control mean = 146.72 seconds
## Test mean = 218.03 seconds
## t-statistic = -9.099
## p-value = 1.39 × 10⁻¹⁹
```

```python
## Conclusion
```

```python
## Because the p-value is substantially smaller than the significance level of 0.05, the null hypothesis is rejected.
## This provides strong statistical evidence that users in the redesigned process (Test group) spent a significantly different amount of time on the confirm step compared with users in the original process (Control group).
## The Test group required approximately 71 additional seconds on average to complete the confirm step (218.03 vs 146.72 seconds). This suggests that although the redesign may have changed user behavior, the confirm stage became noticeably slower and may require further investigation to determine whether additional complexity, hesitation, or usability issues were introduced.
```
